```
for user in users:
    session.sql(f'use role dba_{user};').collect()
    session.sql('use database hol;').collect()
    session.sql(f'use schema schema_{user};').collect()
    
    # the Interactive Warehouse first (it only auto-suspends after 24 hours)
    session.sql(f'drop warehouse if exists std_wh_{user};').collect()
    
    # Interactive Tables
    session.sql('drop table if exists top_sellers;').collect()
    session.sql('drop table if exists customer_recs;').collect()
    session.sql('drop table if exists product_page;').collect()
    
    # source tables
    session.sql('drop table if exists recommendations;').collect()
    session.sql('drop table if exists order_items;').collect()
    session.sql('drop table if exists orders;').collect()
    session.sql('drop table if exists customers;').collect()
    session.sql('drop table if exists products;').collect()
    
    # normal warehouse
    session.sql(f'drop warehouse if exists int_wh_{user};').collect()
```
